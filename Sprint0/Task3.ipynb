Создать базовый класс CreditCard в C#, который будет представлять информацию о кредитных картах. На основе этого класса разработать 3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма. Дополнительно: Необходимо реализовать инкапсуляцию атрибутов с использованием геттеров и сеттеров, а также добавить метод, демонстрирующий взаимодействие объектов между собой.

### Требования к базовому классу CreditCard:

- Атрибуты: Номер карты (CardNumber), Владелец (HolderName), Срок действия (ExpiryDate), Баланс (Balance). Атрибут Balance должен быть инкапсулирован через геттер и сеттер с логикой проверки (баланс не может быть отрицательным).   
- Методы:   
- GetInfo(): метод для получения информации о кредитной карте в виде строки.   
- Pay(decimal amount): метод для списания средств при оплате покупок с проверкой доступного остатка.   
- CheckBalance(): метод для вывода текущего состояния баланса на экран.   
- Transfer(CreditCard targetCard, decimal amount): метод для перевода средств с текущей карты на другую карту (демонстрация взаимодействия объектов).

```
public class CreditCard 
{
    public string CardNumber { get; set; } 
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }

    private decimal _balance; 
    public decimal Balance
    {
        get {return _balance; }
        set {if (value >= 0) {_balance = value; } else {Console.WriteLine("Баланс не может быть отрицательным."); } }
    }
    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = initialBalance;
    }
    public virtual string GetInfo()
    {
        return $"Карта: {CardNumber}, Владелец: {HolderName}, Годен до: {ExpiryDate}";
    }
    public virtual void Pay(decimal amount)
    {
        if (amount <= Balance) { Balance -= amount; Console.WriteLine($"Оплата {amount} прошла успешно."); }
        else { Console.WriteLine("Недостаточно средств."); }
    }
    public virtual void CheckBalance()
    {Console.WriteLine($"Текущий баланс: {Balance}"); }
    public void Transfer(CreditCard targetCard, decimal amount)
    {
        if (amount <= Balance) { Balance -= amount; targetCard.Balance += amount; Console.WriteLine($"Перевод {amount} выполнен."); }
        else {Console.WriteLine("Недостаточно средств для перевода."); }
    }
}

public class GoldCard : CreditCard
{
    public int BonusMiles { get; set; }
    public GoldCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance) : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        BonusMiles = 0;
    }
    public override void Pay(decimal amount)
    {
        if (amount <= Balance) { base.Pay(amount); BonusMiles += 15; Console.WriteLine($"Вам начислено 15 миль. Всего миль: {BonusMiles}"); }
        else { Console.WriteLine("Недостаточно средств."); } 
    }
}
public class PremiumCard : CreditCard
{
    public string SupportAssistant { get; set; }
    public PremiumCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string supportAssistant) : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        SupportAssistant = supportAssistant;
    }
    public override void CheckBalance()
    {
        base.CheckBalance();
        Console.WriteLine($"Ваш персональный ассистент: {SupportAssistant}");
    }

}
public class CorporateCard : CreditCard
{
    public string Company { get; set; }
    public CorporateCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string companyName) : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        Company = companyName;
    }
    public override string GetInfo()
    { 
        return $"{base.GetInfo()}, Компания: {Company}"; 
    }

}

GoldCard myGold = new GoldCard("1111-2222", "Железняк Дамир", "12/26", 5000);
PremiumCard friendCard = new PremiumCard("3333-4444", "Артём Курманов", "10/27", 1000, "Анна");
Console.WriteLine(myGold.GetInfo());
myGold.Pay(1000);
myGold.Transfer(friendCard, 1500);
friendCard.CheckBalance();
friendCard.Balance = -500;




```
