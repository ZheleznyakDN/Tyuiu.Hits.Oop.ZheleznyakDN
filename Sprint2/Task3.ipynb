# Индивидуальный проект

**Название проекта:** Система управления кредитными картами с использованием принципов ООП


**Описание проекта:**
Создать базовый класс `CreditCard` в C#, который представляет информацию о кредитных картах. На основе этого класса разработать 3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма. Дополнительно реализована инкапсуляция атрибутов с использованием геттеров и сеттеров, а также метод, демонстрирующий взаимодействие объектов между собой.

**Требования к базовому классу CreditCard:**
* **Атрибуты:** Номер карты (CardNumber), Владелец (HolderName), Срок действия (ExpiryDate), Баланс (Balance). Атрибут Balance инкапсулирован через геттер и сеттер с логикой проверки (баланс не может быть отрицательным).
* **Методы:**
  * `GetInfo()`: метод для получения информации о кредитной карте в виде строки.
  * `Pay(decimal amount)`: метод для списания средств при оплате покупок с проверкой доступного остатка.
  * `CheckBalance()`: метод для вывода текущего состояния баланса на экран.
  * `Transfer(CreditCard targetCard, decimal amount)`: метод для перевода средств с текущей карты на другую карту.

**Требования к производным классам:**
1. **Золотая карта (GoldCreditCard):** Содержит атрибут Бонусные мили (BonusMiles). Метод `Pay()` переопределен для добавления логики начисления бонусных миль (1% от суммы) при успешной оплате.
2. **Премиальная карта (PremiumCreditCard):** Содержит атрибут Ассистент поддержки (SupportAssistant). Метод `CheckBalance()` переопределен для предоставления возможности обратиться к ассистенту.
3. **Корпоративная карта (CorporateCreditCard):** Содержит атрибут Компания (Company). Метод `GetInfo()` переопределен для включения информации о компании.

**Дополнительное задание:**
Добавьте к существующим классам (базовому и производным) 3-4 атрибута и метода и реализуйте простое, сложное и множественное наследование.

**Реализация дополнительного задания:**
* **Простое наследование:** `GoldCreditCard`, `PremiumCreditCard`, `CorporateCreditCard` напрямую наследуют `CreditCard`.
* **Сложное наследование:** Создан класс `VIPCorporateCard`, который наследуется от производного класса `CorporateCreditCard`.
* **Множественное наследование:** Созданы интерфейсы `ICashback` и `IInsurance`, которые одновременно реализуются в классе `PremiumCreditCard`.
* **Добавленные атрибуты и методы (4 шт):** Свойство `IsBlocked` и `CreditLimit`, методы `BlockCard()` и `UnblockCard()`.

```
using System;

//создаем объекты и вызываем методы
GoldCreditCard myGoldCard = new GoldCreditCard("4276-1111-2222-3333", "Зигфрид Кряквочкин", "12/28", 50000m);
PremiumCreditCard myPremiumCard = new PremiumCreditCard("5555-0000-9999-8888", "Элеонора Шнобель", "10/30", 150000m, "Бромгидрин", 300000m);
VIPCorporateCard myCorpCard = new VIPCorporateCard("1234-5678-9012-3456", "Фантагор Иллюзионов", "05/29", 500000m, "ЗАО 'Межгалактические Рога и Копыта'", "First Class Priority");

//полиморфизма
Console.WriteLine("=== информация о картах ===");
myGoldCard.GetInfo();
myPremiumCard.GetInfo();
myCorpCard.GetInfo();
Console.WriteLine();

Console.WriteLine("=== оплата и начисление бонусов ===");
myGoldCard.Pay(5000m); //переопределенный метод с милями
Console.WriteLine();
myPremiumCard.Pay(10000m); //переопределенный метод с кэшбэком
Console.WriteLine();

Console.WriteLine("=== проверка баланса ===");
myPremiumCard.CheckBalance();
Console.WriteLine();

// демонстрация взаимодействия двух объектов
Console.WriteLine("=== перевод средств ===");
myPremiumCard.Transfer(myGoldCard, 15000m);
myGoldCard.CheckBalance();
Console.WriteLine();

Console.WriteLine("=== дополнительные методы ===");
myPremiumCard.ActivateInsurance();
myGoldCard.BlockCard();
myGoldCard.Pay(1000m); // проверка блокировки: оплата не должна пройти


//объявления интерфейсов и классов

// интерфейсы нужны для реализации множественного наследования (из доп. задания)
public interface ICashback
{
    decimal CashbackPercent { get; set; }
    void ApplyCashback(decimal amount);
}

public interface IInsurance
{
    bool HasTravelInsurance { get; set; }
    void ActivateInsurance();
}

// базовый родительский класс
public class CreditCard
{
    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }

    // инкапсуляция
    private decimal balance;
    public decimal Balance
    {
        get { return balance; }
        set
        {
            if (value < 0)
                throw new ArgumentException("баланс не может быть отрицательным");
            balance = value;
        }
    }

    public bool IsBlocked { get; private set; }

    // конструктор базового класса
    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = initialBalance;
        IsBlocked = false;
    }

    
    public virtual void GetInfo()
    {
        Console.WriteLine($"Карта: {CardNumber} | Владелец: {HolderName} | Срок: {ExpiryDate} | Баланс: {Balance} руб.");
    }

    public virtual bool Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine($"[отказ] карта {CardNumber} заблокирована.");
            return false;
        }
        
        if (Balance >= amount)
        {
            Balance -= amount;
            Console.WriteLine($"[успешно] оплата {amount} руб. прошла. остаток: {Balance} руб.");
            return true;
        }
        else
        {
            Console.WriteLine($"[отказ] недостаточно средств на карте {CardNumber}.");
            return false;
        }
    }

    public virtual void CheckBalance()
    {
        Console.WriteLine($"Текущий баланс карты {CardNumber}: {Balance} руб.");
    }

    // метод принимает другой объект класса CreditCard для перевода денег
    public void Transfer(CreditCard targetCard, decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine("[ошибка перевода] ваша карта заблокирована.");
            return;
        }

        if (this.Balance >= amount)
        {
            this.Balance -= amount;
            targetCard.Balance += amount;
            Console.WriteLine($"[перевод] {amount} руб. успешно переведены на карту {targetCard.CardNumber}.");
        }
        else
        {
            Console.WriteLine("[ошибка перевода] недостаточно средств.");
        }
    }

    public void BlockCard()
    {
        IsBlocked = true;
        Console.WriteLine($"Карта {CardNumber} успешно заблокирована.");
    }

    public void UnblockCard()
    {
        IsBlocked = false;
        Console.WriteLine($"Карта {CardNumber} разблокирована.");
    }
}

// простое наследование от базового класса
public class GoldCreditCard : CreditCard
{
    public decimal BonusMiles { get; private set; }

    public GoldCreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        BonusMiles = 0;
    }

    public override bool Pay(decimal amount)
    {
        bool isSuccess = base.Pay(amount);
        if (isSuccess)
        {
            decimal miles = amount * 0.01m; // 1% миль от суммы
            BonusMiles += miles;
            Console.WriteLine($"[бонус] начислено {miles} миль. всего миль: {BonusMiles}.");
        }
        return isSuccess; // возвращаем результат, прошла ли оплата
    }
}

// множественное наследование
public class PremiumCreditCard : CreditCard, ICashback, IInsurance
{
    public string SupportAssistant { get; set; }
    public decimal CreditLimit { get; set; } 
    public decimal CashbackPercent { get; set; }
    public bool HasTravelInsurance { get; set; }

    public PremiumCreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string assistant, decimal creditLimit)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        SupportAssistant = assistant;
        CreditLimit = creditLimit;
        CashbackPercent = 5.0m; 
        HasTravelInsurance = false;
    }

    public override void CheckBalance()
    {
        base.CheckBalance(); // сначала выведет стандартный баланс
        Console.WriteLine($"[премиум] ваш личный ассистент {SupportAssistant} готов помочь. доступный кредитный лимит: {CreditLimit} руб.");
    }

    // реализация метода из интерфейса ICashback
    public void ApplyCashback(decimal amount)
    {
        decimal cashback = amount * (CashbackPercent / 100);
        Balance += cashback;
        Console.WriteLine($"[кэшбэк] возвращено {cashback} руб. текущий баланс: {Balance} руб.");
    }

    // реализация метода из интерфейса IInsurance
    public void ActivateInsurance()
    {
        HasTravelInsurance = true;
        Console.WriteLine("[страховка] туристическая страховка для путешествий успешно активирована.");
    }

    public override bool Pay(decimal amount)
    {
        bool isSuccess = base.Pay(amount);
        if (isSuccess)
        {
            ApplyCashback(amount);
        }
        return isSuccess;
    }
}

// еще одно простое наследование
public class CorporateCreditCard : CreditCard
{
    public string Company { get; set; }

    public CorporateCreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string company)
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        Company = company;
    }

    public override void GetInfo()
    {
        base.GetInfo(); 
        Console.WriteLine($"Компания-работодатель: {Company}");
    }
}

// сложное наследование
public class VIPCorporateCard : CorporateCreditCard
{
    public string LoungeAccessLevel { get; set; }

    public VIPCorporateCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string company, string loungeLevel)
        : base(cardNumber, holderName, expiryDate, initialBalance, company)
    {
        LoungeAccessLevel = loungeLevel;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Уровень доступа в бизнес-залы: {LoungeAccessLevel}");
    }
}
```
