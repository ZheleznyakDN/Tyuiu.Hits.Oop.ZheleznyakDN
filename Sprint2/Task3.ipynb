Развитие системы управления кредитными картами. Проект включает базовый класс CreditCard, представляющий основную информацию о картах, и производные классы для демонстрации полиморфизма.

### Дополнительное задание: 

К существующим классам добавлено 3 новых атрибута (IsBlocked, BonusPoints, PersonalManager) и 3 новых метода (BlockCard(), AddBonus(), CallManager()).
Реализованы различные типы наследования:   
- Простое наследование: CorporateCard наследуется от CreditCard.
- Сложное (многоуровневое) наследование: Создан новый класс VIPCorporateCard, который наследуется от CorporateCard (иерархия CreditCard -> CorporateCard -> VIPCorporateCard).
- Множественное наследование: В C# реализовано через интерфейсы. Класс GoldCard одновременно наследуется от базового класса CreditCard и реализует интерфейс IBonusSystem.


<h2 style="color:DodgerBlue">Реализация:</h2>

```
using System;
using System.Collections.Generic;

// --- 1. ИСПОЛНЯЕМЫЙ КОД ---

// Воссоздание точного вывода из примера (Простое наследование и базовая логика)
GoldCard card1 = new GoldCard("1111-2222-3333", "Иван Иванов", "12.28", 50000m);
CorporateCard card2 = new CorporateCard("7777-8888-9999", "Петр Петров", "10.29", 500000m, "ООО Инновации");

Console.WriteLine(card1.GetInfo());
card1.CheckBalance();
Console.WriteLine();

Console.WriteLine(card2.GetInfo());
card2.CheckBalance();
Console.WriteLine();

// Демонстрация перевода
card2.Transfer(card1, 25000m);
Console.WriteLine();

card1.CheckBalance();
card2.CheckBalance();
Console.WriteLine(new string('-', 50));


// Демонстрация новых требований (Сложное и Множественное наследование, новые методы)
Console.WriteLine("\n[Демонстрация дополнительного задания]");

// 1. Сложное (многоуровневое) наследование: VIPCorporateCard -> CorporateCard -> CreditCard
VIPCorporateCard vipCard = new VIPCorporateCard("5555-6666-7777", "Анна Смирнова", "11.30", 150000m, "ООО Инновации", "Михаил");
Console.WriteLine(vipCard.GetInfo());
vipCard.CallManager(); // Вызов нового метода №1

// 2. Множественное наследование: GoldCard наследует CreditCard + интерфейс IBonusSystem
card1.Pay(5000m); // При оплате автоматически срабатывает метод AddBonus() (Новый метод №2)

// 3. Работа нового атрибута и метода безопасности базового класса
card2.BlockCard(); // Вызов нового метода №3
card2.Pay(1000m);  // Попытка оплаты заблокированной картой


// --- 2. ИНТЕРФЕЙСЫ И ОБЪЯВЛЕНИЕ КЛАССОВ ---

// Интерфейс для реализации множественного наследования
public interface IBonusSystem
{
    // Новый атрибут №1 (через свойство интерфейса)
    int BonusPoints { get; set; }
    // Новый метод №1 (в интерфейсе)
    void AddBonus(decimal amount);
}

// Базовый класс
public class CreditCard
{
    private decimal _balance;

    public string CardNumber { get; set; }
    public string HolderName { get; set; }
    public string ExpiryDate { get; set; }
    
    // Новый атрибут №2
    public bool IsBlocked { get; set; } = false;

    public decimal Balance
    {
        get { return _balance; }
        set { _balance = value >= 0 ? value : 0; }
    }

    public CreditCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance)
    {
        CardNumber = cardNumber;
        HolderName = holderName;
        ExpiryDate = expiryDate;
        Balance = initialBalance;
    }

    public virtual string GetInfo()
    {
        return $"Карта: {CardNumber} | Владелец: {HolderName} | Срок: {ExpiryDate}";
    }

    // Изменен под формат вывода из примера
    public virtual void CheckBalance()
    {
        Console.WriteLine($"Текущий баланс: {Math.Round(Balance)} руб.");
    }

    // Изменен под формат вывода из примера
    public void Transfer(CreditCard targetCard, decimal amount)
    {
        if (!IsBlocked && !targetCard.IsBlocked && amount > 0 && Balance >= amount)
        {
            Balance -= amount;
            targetCard.Balance += amount;
            Console.WriteLine($"Перевод {Math.Round(amount)} руб. с карты {CardNumber} на карту {targetCard.CardNumber} выполнен успешно.");
        }
        else
        {
            Console.WriteLine($"Ошибка перевода. Проверьте баланс или статус блокировки карт.");
        }
    }

    public virtual void Pay(decimal amount)
    {
        if (IsBlocked)
        {
            Console.WriteLine($"[Отказ] Карта {CardNumber} заблокирована.");
            return;
        }

        if (amount > 0 && Balance >= amount)
        {
            Balance -= amount;
            Console.WriteLine($"Оплата {Math.Round(amount)} руб. успешно проведена.");
        }
    }

    // Новый метод №2
    public void BlockCard()
    {
        IsBlocked = true;
        Console.WriteLine($"[Безопасность] Карта {CardNumber} была заблокирована.");
    }
}

// ПРОСТОЕ И МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ: Класс наследует CreditCard и интерфейс IBonusSystem
public class GoldCard : CreditCard, IBonusSystem
{
    // Реализация нового атрибута №1 из интерфейса
    public int BonusPoints { get; set; } = 0;

    public GoldCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance) 
        : base(cardNumber, holderName, expiryDate, initialBalance) { }

    // Реализация нового метода №1 из интерфейса
    public void AddBonus(decimal amount)
    {
        int earned = (int)(amount / 100);
        BonusPoints += earned;
        Console.WriteLine($"[Бонус] Начислено {earned} баллов. Баланс баллов: {BonusPoints}.");
    }

    public override void Pay(decimal amount)
    {
        if (!IsBlocked && Balance >= amount)
        {
            base.Pay(amount);
            AddBonus(amount); // Вызываем начисление бонусов при успешной оплате
        }
        else
        {
            base.Pay(amount); // Передаем в базовый метод для вывода ошибки
        }
    }
}

// ПРОСТОЕ НАСЛЕДОВАНИЕ
public class CorporateCard : CreditCard
{
    public string Company { get; set; }

    public CorporateCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string company) 
        : base(cardNumber, holderName, expiryDate, initialBalance)
    {
        Company = company;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $" | Компания: {Company}";
    }
}

// СЛОЖНОЕ (МНОГОУРОВНЕВОЕ) НАСЛЕДОВАНИЕ: VIPCorporateCard -> CorporateCard -> CreditCard
public class VIPCorporateCard : CorporateCard
{
    // Новый атрибут №3
    public string PersonalManager { get; set; }

    public VIPCorporateCard(string cardNumber, string holderName, string expiryDate, decimal initialBalance, string company, string personalManager) 
        : base(cardNumber, holderName, expiryDate, initialBalance, company)
    {
        PersonalManager = personalManager;
    }

    // Новый метод №3
    public void CallManager()
    {
        Console.WriteLine($"[VIP-Сервис] Осуществляется звонок вашему персональному менеджеру: {PersonalManager}...");
    }
}
```
